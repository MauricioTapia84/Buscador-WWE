# 📊 Análisis Exploratorio de Datos (EDA) - Buscador WWE
Este notebook tiene como objetivo analizar las distribuciones y correlaciones en el dataset procesado (`wrestling_clean.csv`) ANTES de entrenar los modelos, para prevenir Data Leakage (fuga de datos).

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Configuraciones visuales
sns.set_theme(style="whitegrid")

df = pd.read_csv('../data/processed/wrestling_clean.csv')
df.head()
```

### Correlación entre Victorias Totales y Posibilidad de ser Campeón

```python
plt.figure(figsize=(8,5))
sns.boxplot(x='is_champion', y='total_wins', data=df)
plt.title('Distribución de Victorias Totales vs Campeón')
plt.show()
```

### Desbalance de Clases

```python
ax = sns.countplot(x='is_champion', data=df)
plt.title('Distribución de Clases (Campeón vs No Campeón)')
for p in ax.patches:
    ax.annotate(f'{p.get_height()}', (p.get_x()+0.3, p.get_height()+10))
plt.show()
```
